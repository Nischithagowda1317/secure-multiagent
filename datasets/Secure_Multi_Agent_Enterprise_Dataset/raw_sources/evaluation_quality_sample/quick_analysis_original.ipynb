# Quick Analysis - Free Kaggle Sample
This notebook loads the free sample and runs a small starter analysis.

```python
import pandas as pd
from pathlib import Path
root = Path('..') / 'sample_output'
daily_evaluation_metrics = pd.read_csv(root / 'daily_evaluation_metrics.csv')
reviewer_decisions = pd.read_csv(root / 'reviewer_decisions.csv')
daily_evaluation_metrics.shape, reviewer_decisions.shape
```

```python
reviewer_decisions.head()
```

```python
joined = reviewer_decisions.merge(daily_evaluation_metrics, on='id', how='left')
joined.head()
```

```python
joined.groupby('review_id')['human_score'].agg(['count','mean']).sort_values('count', ascending=False)
```
